```python
import pandas as pd
import os
data_path = "/content/AI-Fraud-Detection/data"

# Load the main transaction dataset
transactions = pd.read_csv(
    os.path.join(data_path, "transactions.csv")
)

# Load supporting datasets
accounts = pd.read_csv(
    os.path.join(data_path, "account_profiles.csv")
)

network = pd.read_csv(
    os.path.join(data_path, "network_edges.csv")
)

time_stats = pd.read_csv(
    os.path.join(data_path, "time_series_stats.csv")
)

# Display dataset dimensions
print("TRANSACTIONS:", transactions.shape)
print("ACCOUNT PROFILES:", accounts.shape)
print("NETWORK EDGES:", network.shape)
print("TIME SERIES:", time_stats.shape)

print("\nTRANSACTION COLUMNS:")
print(transactions.columns.tolist())

print("\nFIRST 5 TRANSACTIONS:")
display(transactions.head())

print("\nDATA TYPES:")
print(transactions.dtypes)

# Check the fraud patterns file, if it is included
fraud_path = os.path.join(data_path, "fraud_patterns.csv")
if os.path.exists(fraud_path):
    print("\nFRAUD PATTERNS FILE SIZE:",
          os.path.getsize(fraud_path), "bytes")
else:
    print("\nfraud_patterns.csv is not present in the data folder.")
```

```python

import pandas as pd
import numpy as np

# 1. Fraud and legitimate transaction counts
print("CLASS DISTRIBUTION")
print(transactions["is_fraud"].value_counts().sort_index())

fraud_counts = transactions["is_fraud"].value_counts()
total = len(transactions)

print("\nCLASS PERCENTAGES")
print((fraud_counts / total * 100).round(2))

print("\nFraud rate:",
      round(transactions["is_fraud"].mean() * 100, 3), "%")

# 2. Missing values
print("\nMISSING VALUES")
missing = transactions.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))

# 3. Duplicate records
print("\nDUPLICATE CHECK")
print("Duplicate rows:", transactions.duplicated().sum())
print("Duplicate transaction IDs:",
      transactions["transaction_id"].duplicated().sum())

# 4. Timestamp range
transactions["timestamp"] = pd.to_datetime(
    transactions["timestamp"], errors="coerce"
)

print("\nTIMESTAMP RANGE")
print("Earliest:", transactions["timestamp"].min())
print("Latest:", transactions["timestamp"].max())
print("Invalid timestamps:",
      transactions["timestamp"].isna().sum())

# 5. Fraud pattern inspection
print("\nFRAUD PATTERN VALUES")
print(transactions["fraud_pattern"].value_counts(dropna=False))

# 6. Fraud distribution by month
print("\nMONTHLY FRAUD COUNTS")
monthly = transactions.groupby(
    transactions["timestamp"].dt.to_period("M")
)["is_fraud"].agg(["count", "sum"])

monthly["fraud_percentage"] = (
    monthly["sum"] / monthly["count"] * 100
).round(2)

display(monthly)
```

```python
import pandas as pd
import numpy as np

df = transactions.copy()

# Extract useful time features
df["timestamp"] = pd.to_datetime(df["timestamp"])

df["year"] = df["timestamp"].dt.year
df["month"] = df["timestamp"].dt.month
df["day"] = df["timestamp"].dt.day

# Define target
y = df["is_fraud"].astype("int8")

# Columns that must NOT be used as model inputs
drop_columns = [
    "transaction_id",
    "account_id",
    "timestamp",
    "is_fraud",
    "fraud_pattern"
]

# Remove identifiers and target leakage columns
X = df.drop(columns=drop_columns)

# Show feature information
print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nData types:")
print(X.dtypes)

print("\nMissing values:")
print(X.isnull().sum().sum())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Create masks based on transaction year and month
train_mask = X["year"] <= 2023

val_mask = (
    (X["year"] == 2024) &
    (X["month"] <= 6)
)

test_mask = (
    (X["year"] == 2024) &
    (X["month"] >= 7)
)

# Split features and labels
X_train = X.loc[train_mask].copy()
X_val = X.loc[val_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_val = y.loc[val_mask].copy()
y_test = y.loc[test_mask].copy()

# Display split sizes and class distributions
for name, X_part, y_part in [
    ("TRAIN", X_train, y_train),
    ("VALIDATION", X_val, y_val),
    ("TEST", X_test, y_test)
]:
    print(f"\n{name}")
    print("Rows:", len(X_part))
    print("Legitimate:", int((y_part == 0).sum()))
    print("Fraudulent:", int((y_part == 1).sum()))
    print("Fraud percentage:",
          round(y_part.mean() * 100, 3), "%")

# Confirm no overlapping rows
assert len(X_train) + len(X_val) + len(X_test) == len(X)

print("\nChronological split completed successfully!")
```

```python
from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_cols = [
    "merchant_category",
    "merchant_country",
    "device_type",
    "mcc_code"
]

# Identify numerical columns
numerical_cols = [
    col for col in X_train.columns
    if col not in categorical_cols
]

print("Categorical columns:", categorical_cols)
print("Numerical columns:", len(numerical_cols))

# Create encoder
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Fit ONLY on training data
X_train_cat = encoder.fit_transform(
    X_train[categorical_cols].astype(str)
)

# Transform validation and test
X_val_cat = encoder.transform(
    X_val[categorical_cols].astype(str)
)

X_test_cat = encoder.transform(
    X_test[categorical_cols].astype(str)
)

# Extract numerical features
X_train_num = X_train[numerical_cols].to_numpy(dtype=np.float32)
X_val_num = X_val[numerical_cols].to_numpy(dtype=np.float32)
X_test_num = X_test[numerical_cols].to_numpy(dtype=np.float32)

# Combine numerical and encoded categorical features
X_train_encoded = np.hstack([X_train_num, X_train_cat]).astype(np.float32)
X_val_encoded = np.hstack([X_val_num, X_val_cat]).astype(np.float32)
X_test_encoded = np.hstack([X_test_num, X_test_cat]).astype(np.float32)

print("\nEncoded shapes:")
print("Training:", X_train_encoded.shape)
print("Validation:", X_val_encoded.shape)
print("Testing:", X_test_encoded.shape)

print("\nEncoding completed successfully!")
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

scaler = StandardScaler()

# Fit only on training data
X_train_num_scaled = scaler.fit_transform(X_train_num)

# Apply the same scaler to validation and test
X_val_num_scaled = scaler.transform(X_val_num)
X_test_num_scaled = scaler.transform(X_test_num)

# Combine scaled numerical and categorical features
X_train_final = np.hstack([
    X_train_num_scaled, X_train_cat
]).astype(np.float32)

X_val_final = np.hstack([
    X_val_num_scaled, X_val_cat
]).astype(np.float32)

X_test_final = np.hstack([
    X_test_num_scaled, X_test_cat
]).astype(np.float32)

print("Final feature shapes:")
print("Training:", X_train_final.shape)
print("Validation:", X_val_final.shape)
print("Testing:", X_test_final.shape)

print("\nChecking numerical scaling:")
print("Training numerical mean:",
      np.round(X_train_num_scaled.mean(axis=0)[:5], 3))
print("Training numerical standard deviation:",
      np.round(X_train_num_scaled.std(axis=0)[:5], 3))

print("\nScaling completed successfully!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Train the model
lr_model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42,
    solver="lbfgs"
)

print("Training Logistic Regression...")

lr_model.fit(X_train_final, y_train)

print("Training completed!")

# Predict validation probabilities
val_scores_lr = lr_model.predict_proba(X_val_final)[:, 1]

# Default threshold
val_pred_lr = (val_scores_lr >= 0.5).astype(int)

# Evaluate validation performance
print("\n--- LOGISTIC REGRESSION VALIDATION RESULTS ---")

print("PR-AUC:",
      round(average_precision_score(y_val, val_scores_lr), 4))

print("Precision:",
      round(precision_score(y_val, val_pred_lr, zero_division=0), 4))

print("Recall:",
      round(recall_score(y_val, val_pred_lr, zero_division=0), 4))

print("F1-score:",
      round(f1_score(y_val, val_pred_lr, zero_division=0), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, val_pred_lr))

print("\nClassification Report:")
print(classification_report(
    y_val, val_pred_lr, zero_division=0
))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_recall_curve

def analyze_thresholds(y_true, scores, model_name):
    precision, recall, thresholds = precision_recall_curve(y_true, scores)

    # The final precision/recall point has no corresponding threshold.
    precision = precision[:-1]
    recall = recall[:-1]

    f1 = 2 * precision * recall / (precision + recall + 1e-10)

    best_f1_idx = np.argmax(f1)

    # Find the highest-precision threshold that achieves at least 90% recall.
    eligible = np.where(recall >= 0.90)[0]

    if len(eligible) > 0:
        best_recall_idx = eligible[np.argmax(precision[eligible])]
    else:
        best_recall_idx = None

    print(f"\n--- {model_name} ---")
    print("Best F1 threshold:", round(thresholds[best_f1_idx], 4))
    print("Precision:", round(precision[best_f1_idx], 4))
    print("Recall:", round(recall[best_f1_idx], 4))
    print("F1-score:", round(f1[best_f1_idx], 4))

    if best_recall_idx is not None:
        print("\nBest precision with at least 90% recall:")
        print("Threshold:", round(thresholds[best_recall_idx], 4))
        print("Precision:", round(precision[best_recall_idx], 4))
        print("Recall:", round(recall[best_recall_idx], 4))
        print("F1-score:", round(f1[best_recall_idx], 4))
    else:
        print("\nNo threshold achieved 90% recall.")

analyze_thresholds(
    y_val,
    val_scores_rf,
    "Balanced Random Forest"
)

analyze_thresholds(
    y_val,
    val_scores_unweighted,
    "Unweighted Random Forest"
)
```

```python
import gradio as gr
import pandas as pd
import numpy as np

def predict_transaction(
    hour_of_day,
    day_of_week,
    is_weekend,
    amount,
    merchant_category,
    merchant_country,
    card_present,
    device_type,
    device_known,
    ip_risk_score,
    is_foreign_txn,
    time_since_last_s,
    velocity_1h,
    amount_vs_avg_ratio,
    account_age_days,
    has_2fa,
    credit_limit,
    year,
    month,
    day,
    mcc_code
):
    # Create a single-row input using the original feature names
    row = {
        "hour_of_day": hour_of_day,
        "day_of_week": day_of_week,
        "is_weekend": is_weekend,
        "amount": amount,
        "merchant_category": merchant_category,
        "merchant_country": merchant_country,
        "mcc_code": str(mcc_code),
        "card_present": card_present,
        "device_type": device_type,
        "device_known": device_known,
        "ip_risk_score": ip_risk_score,
        "is_foreign_txn": is_foreign_txn,
        "time_since_last_s": time_since_last_s,
        "velocity_1h": velocity_1h,
        "amount_vs_avg_ratio": amount_vs_avg_ratio,
        "account_age_days": account_age_days,
        "has_2fa": has_2fa,
        "credit_limit": credit_limit,
        "year": year,
        "month": month,
        "day": day
    }

    input_df = pd.DataFrame([row])

    # Apply the exact training preprocessing
    numeric_data = input_df[numerical_cols].to_numpy()
    scaled_data = scaler.transform(numeric_data)

    categorical_data = input_df[categorical_cols].copy()
    categorical_data["mcc_code"] = categorical_data["mcc_code"].astype(str)
    encoded_data = encoder.transform(categorical_data)

    final_input = np.hstack([scaled_data, encoded_data])

    # Predict fraud probability
    probability = float(rf_unweighted.predict_proba(final_input)[0, 1])
    prediction = int(probability >= final_threshold)

    result = "FRAUDULENT TRANSACTION" if prediction == 1 else "LEGITIMATE TRANSACTION"

    return result, f"{probability * 100:.2f}%"

print("Prediction function created successfully!")
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    average_precision_score, precision_recall_curve,
    precision_score, recall_score, f1_score, confusion_matrix
)

# Use the already-created chronological split and calendar-free features.
Xtr = X_train_final_no_calendar
Xva = X_val_final_no_calendar
Xte = X_test_final_no_calendar
ytr = np.asarray(y_train).astype(int)
yva = np.asarray(y_val).astype(int)
yte = np.asarray(y_test).astype(int)

# Guard against accidentally comparing different feature representations.
assert Xtr.shape[0] == len(ytr)
assert Xva.shape[0] == len(yva)
assert Xte.shape[0] == len(yte)
assert Xtr.shape[1] == Xva.shape[1] == Xte.shape[1]

# Define a threshold using validation data only.
def choose_validation_thresholds(y_true, scores, target_recall=0.90):
    precision, recall, thresholds = precision_recall_curve(y_true, scores)
    # precision/recall have one more element than thresholds.
    f1_values = 2 * precision[:-1] * recall[:-1] / (
        precision[:-1] + recall[:-1] + 1e-12
    )
    f1_idx = int(np.nanargmax(f1_values))
    f1_threshold = float(thresholds[f1_idx])

    eligible = np.flatnonzero(recall[:-1] >= target_recall)
    if len(eligible) == 0:
        # The lowest score threshold should ordinarily permit recall >= target.
        # Keep this explicit fallback in case of unusual estimator outputs.
        recall_threshold = float(np.min(scores))
    else:
        best_precision = np.max(precision[eligible])
        tied = eligible[np.isclose(precision[eligible], best_precision)]
        # On a precision tie, use the highest threshold (fewer alerts).
        recall_idx = tied[np.argmax(thresholds[tied])]
        recall_threshold = float(thresholds[recall_idx])
    return f1_threshold, recall_threshold


def score_at_threshold(y_true, scores, threshold):
    pred = (scores >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
    return {
        "precision": float(precision_score(y_true, pred, zero_division=0)),
        "recall": float(recall_score(y_true, pred, zero_division=0)),
        "f1": float(f1_score(y_true, pred, zero_division=0)),
        "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
        "alerts": int(pred.sum())
    }

# Keep seed 42 as the direct weighted/unweighted comparison for LR and DT.
# Repeat RF with seeds 42, 1, and 2 to quantify random-seed variability.
experiments = []
model_jobs = []

for weighting, class_weight in [("weighted", "balanced"), ("unweighted", None)]:
    model_jobs.append({
        "model_family": "Logistic Regression",
        "weighting": weighting,
        "class_weight": class_weight,
        "seed": 42,
        "estimator": LogisticRegression(
            class_weight=class_weight, max_iter=1000,
            solver="lbfgs", random_state=42
        )
    })
    model_jobs.append({
        "model_family": "Decision Tree",
        "weighting": weighting,
        "class_weight": class_weight,
        "seed": 42,
        "estimator": DecisionTreeClassifier(
            class_weight=class_weight, max_depth=12,
            min_samples_leaf=50, random_state=42
        )
    })
    for seed in [42, 1, 2]:
        model_jobs.append({
            "model_family": "Random Forest",
            "weighting": weighting,
            "class_weight": "balanced_subsample" if weighting == "weighted" else None,
            "seed": seed,
            "estimator": RandomForestClassifier(
                n_estimators=200, max_depth=16, min_samples_leaf=20,
                class_weight="balanced_subsample" if weighting == "weighted" else None,
                random_state=seed, n_jobs=-1
            )
        })

fitted_experiment_models = {}

for job_no, job in enumerate(model_jobs, start=1):
    family = job["model_family"]
    weighting = job["weighting"]
    seed = job["seed"]
    label = f"{family} | {weighting} | seed={seed}"
    print(f"\n[{job_no}/{len(model_jobs)}] Fitting {label}")

    model = job["estimator"]
    model.fit(Xtr, ytr)
    fitted_experiment_models[(family, weighting, seed)] = model

    val_scores = model.predict_proba(Xva)[:, 1]
    test_scores = model.predict_proba(Xte)[:, 1]
    val_ap = average_precision_score(yva, val_scores)
    test_ap = average_precision_score(yte, test_scores)
    threshold_f1, threshold_recall90 = choose_validation_thresholds(yva, val_scores, 0.90)

    row = {
        "model_family": family,
        "weighting": weighting,
        "class_weight": str(job["class_weight"]),
        "seed": seed,
        "n_features": int(Xtr.shape[1]),
        "train_rows": int(Xtr.shape[0]),
        "validation_rows": int(Xva.shape[0]),
        "test_rows": int(Xte.shape[0]),
        "validation_PR_AUC": float(val_ap),
        "test_PR_AUC": float(test_ap),
        "threshold_max_validation_F1": threshold_f1,
        "threshold_validation_recall_at_least_90pct": threshold_recall90,
    }

    for rule, threshold in [
        ("max_F1", threshold_f1),
        ("precision_at_recall90", threshold_recall90)
    ]:
        val_metrics = score_at_threshold(yva, val_scores, threshold)
        test_metrics = score_at_threshold(yte, test_scores, threshold)
        for metric, value in val_metrics.items():
            row[f"validation_{rule}_{metric}"] = value
        for metric, value in test_metrics.items():
            row[f"test_{rule}_{metric}"] = value

    experiments.append(row)

comparison_df = pd.DataFrame(experiments).sort_values(
    ["model_family", "weighting", "seed"]
).reset_index(drop=True)

print("\n========== WEIGHTED VS UNWEIGHTED + SEED RESULTS ==========")
print("PR-AUC is threshold-independent. Thresholds were selected from validation only.")
print(comparison_df[[
    "model_family", "weighting", "seed", "validation_PR_AUC", "test_PR_AUC",
    "threshold_max_validation_F1", "validation_max_F1_precision",
    "validation_max_F1_recall", "validation_max_F1_f1",
    "threshold_validation_recall_at_least_90pct",
    "validation_precision_at_recall90_precision",
    "validation_precision_at_recall90_recall",
    "validation_precision_at_recall90_f1",
    "test_max_F1_precision", "test_max_F1_recall", "test_max_F1_f1",
    "test_precision_at_recall90_precision",
    "test_precision_at_recall90_recall",
    "test_precision_at_recall90_f1"
]].round(4).to_string(index=False))

# Save only new files; existing artifacts and model files are not overwritten.
if "improvement_dir" not in globals():
    improvement_dir = "/content/drive/MyDrive/Financial_Fraud_Project/results/improvement_experiments"
os.makedirs(improvement_dir, exist_ok=True)

comparison_path = os.path.join(improvement_dir, "weighted_unweighted_model_comparison.csv")
comparison_df.to_csv(comparison_path, index=False)

rf_seed_df = comparison_df[comparison_df["model_family"] == "Random Forest"].copy()
rf_seed_path = os.path.join(improvement_dir, "random_forest_seed_stability.csv")
rf_seed_df.to_csv(rf_seed_path, index=False)

# Concise seed stability summary, based on validation PR-AUC (not test performance).
seed_summary = rf_seed_df.groupby("weighting").agg(
    validation_PR_AUC_mean=("validation_PR_AUC", "mean"),
    validation_PR_AUC_std=("validation_PR_AUC", "std"),
    validation_PR_AUC_min=("validation_PR_AUC", "min"),
    validation_PR_AUC_max=("validation_PR_AUC", "max"),
    validation_max_F1_mean=("validation_max_F1_f1", "mean"),
    validation_max_F1_std=("validation_max_F1_f1", "std"),
).reset_index()
seed_summary_path = os.path.join(improvement_dir, "random_forest_seed_summary.csv")
seed_summary.to_csv(seed_summary_path, index=False)

print("\n========== RANDOM FOREST SEED STABILITY (VALIDATION) ==========")
print(seed_summary.round(4).to_string(index=False))
print("\nSaved new files:")
print(comparison_path)
print(rf_seed_path)
print(seed_summary_path)
```

```python
# 2. Cost comparison: flag nothing vs F1-optimal vs cost-optimal threshold
import numpy as np
import pandas as pd
from sklearn.metrics import precision_recall_curve, confusion_matrix

FP_COST = 1.0
FN_COST = 10.0  # hypothetical cost ratio; document/replace if approved

y_val_cost = np.asarray(y_val).astype(int)
p_val_cost = np.asarray(y_val_proba_no_calendar)
precision_v, recall_v, thresholds_v = precision_recall_curve(y_val_cost, p_val_cost)
f1_v = 2 * precision_v[:-1] * recall_v[:-1] / (precision_v[:-1] + recall_v[:-1] + 1e-12)
f1_threshold_cost = float(thresholds_v[int(np.nanargmax(f1_v))])

# Cost threshold selected using validation labels only.
validation_costs = []
for t in thresholds_v:
    pv = (p_val_cost >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_val_cost, pv, labels=[0, 1]).ravel()
    validation_costs.append(fp * FP_COST + fn * FN_COST)
cost_threshold_cost = float(thresholds_v[int(np.argmin(validation_costs))])

y_test_cost = np.asarray(y_test).astype(int)
p_test_cost = np.asarray(y_test_proba_no_calendar)
test_scenarios = [
    ("Flag nothing", None),
    ("F1-optimal threshold (validation)", f1_threshold_cost),
    ("Cost-optimal threshold (validation)", cost_threshold_cost)
]
cost_rows = []
for scenario, threshold in test_scenarios:
    pred = np.zeros_like(y_test_cost) if threshold is None else (p_test_cost >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test_cost, pred, labels=[0, 1]).ravel()
    total_cost = float(fp * FP_COST + fn * FN_COST)
    cost_rows.append({
        "scenario": scenario, "threshold": threshold,
        "TP": int(tp), "FP": int(fp), "FN": int(fn), "TN": int(tn),
        "alerts": int(pred.sum()), "cost_FP_unit": FP_COST, "cost_FN_unit": FN_COST,
        "total_cost": total_cost
    })

cost_comparison_df = pd.DataFrame(cost_rows)
baseline_cost = float(cost_comparison_df.loc[0, "total_cost"])
cost_comparison_df["cost_saved_vs_flag_nothing"] = baseline_cost - cost_comparison_df["total_cost"]
cost_comparison_df["percent_cost_saved_vs_flag_nothing"] = (
    100 * cost_comparison_df["cost_saved_vs_flag_nothing"] / baseline_cost if baseline_cost else np.nan
)
print("F1 threshold (validation):", f1_threshold_cost)
print("Cost-optimal threshold (validation):", cost_threshold_cost)
display(cost_comparison_df.round(4))
cost_comparison_df.to_csv(os.path.join(improvement_dir, "cost_comparison_flag_none_f1_cost_optimal.csv"), index=False)
```
